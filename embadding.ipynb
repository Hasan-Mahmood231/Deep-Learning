```python
# import tensorflow as tf
# import keras
# from keras import layers 
# from tf.keras.preprocessing.text import Tokenizer

import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers 
from tensorflow.keras.preprocessing.text import Tokenizer
```

```python
sentences = [
    "the sun is bright",
    "the sky is blue",
    "we feel very happy",
    "today is a sunny day",
    "warm weather feels nice",
    "the golden sun shines",
    "dark clouds bring rain",
    "the storm is cold",
    "i feel very sad",
    "it is raining today",
    "the grey sky is gloomy",
    "cold rain falls down"
]
```

```python
tokenizer = Tokenizer()
tokenizer.fit_on_texts(sentences)
```

```python
len(tokenizer.word_index)
```

```python
sequences = tokenizer.texts_to_sequences(sentences)
sequences
```

```python
from keras.utils import pad_sequences
sequences = pad_sequences(sequences, padding='post')
sequences
```

```python
from keras.models import Sequential
model = Sequential()
model.add(layers.Embedding(input_dim=20, output_dim=8, input_length=6))
model.summary()
```

```python
model.compile(optimizer='adam', loss='accuracy')
```
